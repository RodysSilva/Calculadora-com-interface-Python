```python
#importar bibliotecas
import tkinter as tk
```

```python
#Funções
def click(num):
    entry.insert(tk.END, num)

def clear():
    entry.delete(0, tk.END)

def calculate():
    try:
        result=eval(entry.get)
        entry.delete(0, tk.END)
        entry.insert(tk.END, str(result))
    except:
        entry.delete(0, tk.END)
        entry.insert(tk.END, "Erro")
```

```python
#Criando interface
root=tk.Tk()
root.title("Calculadora Simples")
```

```python
#Cria campo de entrada e resultados
entry=tk.Entry(root, width=20, borderwidth=5, font=("Arial", 14))
entry.grid(row=0, column=0, columnspan=4)
```

```python
#Lista com rotulos dos botões
buttons=[
    '7', '8', '9', '+',
    '4', '5', '6', '-',
    '1', '2', '3', '*',
    '0', 'C', '=', '/'
]   

row, col=1, 0
```

```python
#Loop que cria e posiciona botoões 
for b in buttons:
    if b=="c":
        action=clear
    elif b=="=":
        action=calculate
    else:
        action=lambda x=b: click(x)

    btn=tk.Button(root, text=b, width=5, height=2, command=action)
    btn.grid(row=row, column=col)

    col+=1
    if col>3:
        col=0
        row+=1

root.mainloop()
```
